### Loading the image with OpenCV

```python
import cv2
import matplotlib.pyplot as plt

img1 = cv2.imread('/home/aryan/code/Computer_vision/Media/img1.jpg')

# 1. Convert BGR to RGB and store the image array in img1_rgb
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

# 2. Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,12))

# 3. Display images on each subplot
ax1.imshow(img1)
ax1.set_title('Original BGR')

ax2.imshow(img1_rgb)
ax2.set_title('Converted RGB')

plt.tight_layout()
plt.show()
```

### Resizing

```python
b = cv2.resize(img1_rgb, (50,50))
plt.imshow(b)
print(img1_rgb.shape)
print(b.shape)  ## Down scaled the image to 50x50 pixels
```

### Grayscale

```python
img1_grey=cv2.cvtColor(img1_rgb,cv2.COLOR_RGB2GRAY)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15,15))

ax1.imshow(img1_rgb)
ax1.set_title('RGB')

ax2.imshow(img1_grey, cmap='grey')
ax2.set_title("Grey")

plt.tight_layout()
plt.show()

```

## Normalize

```python
import numpy as np 
v = np.array(img1_rgb)  ## CV already returns a nparray

b = v / 255   ## dividing each pixel with 255 (max intensity value)

plt.imshow(b)
print('This is the median value for original image: ',np.median(v))
print('This is the median value for normalized image: ',np.median(b))
```
