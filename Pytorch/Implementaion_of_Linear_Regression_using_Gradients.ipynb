```python
import torch
import numpy as np
```

## This is the implementaion using Torch with Autograd

```python
def linear_regression_gradient_descent(X, y, alpha, iterations) -> torch.Tensor:
    """
    Perform linear regression using gradient descent with PyTorch autograd.

    Args:
        X: Feature matrix (m, n) - can be tensor or array-like
        y: Target vector (m,) - can be tensor or array-like
        alpha: Learning rate
        iterations: Number of gradient descent iterations

    Returns:
        Learned weights as a 1D tensor of shape (n,)
    """
    X_t = torch.as_tensor(X, dtype=torch.float32)
    y_t = torch.as_tensor(y, dtype=torch.float32).reshape(-1, 1)
    m, n = X_t.shape
    theta = torch.zeros((n, 1), requires_grad=True)     ## Auto grad keeps the track of the variable to perform the gradients


    for _ in range(iterations):

      score = X_t @ theta  ## shape (m,n) @ (n, 1)

      loss = torch.mean((score - y_t) ** 2) / 2   ## Here we are calculating the loss func and the / 2 is for the backprop

      loss.backward()     # Autograd computes d(loss)/d(theta)

      with torch.no_grad():             ## "Don't track this weight update as part of the computational graph."
        theta -= alpha * theta.grad     ## update the theta


      theta.grad.zero_()            ## Remove the current stored grad value so that it does not accumulate in the future


    return theta.detach().flatten()  ## detach removes the autograd from the variable so its no longer involved
```

```python
X = torch.tensor([[1.0, 1.0], [1.0, 2.0], [1.0, 3.0]])
y = torch.tensor([3.0, 5.0, 7.0])
linear_regression_gradient_descent(X, y, 0.1, 1000)

```

## This is the implementation of the same thing but in numpy

```python
def linear_regression_gradient_descent(X: np.ndarray, y: np.ndarray, alpha: float, iterations: int) -> np.ndarray:
    """
    Perform linear regression using gradient descent.

    Args:
        X: Feature matrix of shape (m, n) where first column is all ones (for intercept)
        y: Target vector of shape (m,)
        alpha: Learning rate
        iterations: Number of gradient descent iterations

    Returns:
        Learned weights as a 1D array of shape (n,)
    """
    m, n = X.shape
    y = y.reshape(-1, 1)  # Ensure y is a column vector
    theta = np.zeros((n, 1))  # Initialize weights to zeros


    for _ in range(iterations):

      score = X @ theta  ## [m, n] @ [n, 1]

      error = (score - y)      ## we are not using mean here as we know the derivative so we can calculate that manually [m, 1]

      grad = (X.T @ error) / m     ## m is for the avg sum

      theta = theta - alpha * grad

    return theta.flatten()
```

```python
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([3, 5, 7])
alpha = 0.1
iterations = 1000
linear_regression_gradient_descent(X, y, alpha, iterations)
```
