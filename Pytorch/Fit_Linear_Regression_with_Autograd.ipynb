## Fit Linear Regression with Autograd lab

```python
import torch
import numpy
```

```python
x = torch.rand(160, 3)
y = torch.rand(160)
lr = 0.1
steps = 500

n,d = x.shape

w = torch.rand(d, requires_grad=True)
b = torch.rand(1, requires_grad=True)

for _ in range(steps):

  score = (x @ w) + b

  loss = torch.mean((score - y) ** 2) / 2     ## Calculating the MSE

  loss.backward()   ## Calulating the gradients


  with torch.no_grad():
    w -= lr * w.grad    ## Here w is getting a inplace replacement with w-= if we do w = w - .. that will create a new varible w that wont have the gradient graph attached causing error
    b -= lr * b.grad    ## Same as above
  w.grad.zero_()
  b.grad.zero_()

print(w.detach(), b.detach())     ## Returns with gradient graph removed
```
