## Classic CV Filters with torch and CV

```python
import cv2
import torch
import matplotlib.pyplot as plt
```

- Here i have avoided the border pixels in the range loop as this is without any padding 

```python
from torch import from_numpy
from cv2 import COLOR_BGR2RGB
## Set your image
b = cv2.imread('/home/aryan/code/Computer_vision/Media/img3.webp')
## Set your kernel
kernel = torch.tensor([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=torch.float32) / 16

```

```python


img = cv2.cvtColor(b ,cv2.COLOR_BGR2GRAY)       
img = cv2.resize(img, (250, 250))            ## Resize the image for faster computation 
img = torch.tensor(img, dtype=torch.float32)
result = torch.zeros_like(img)              ## Creates the matrix of zeros of the same size as img


def OPS(x, y, kernel, result, img):
    for i in range(1, x-1):
        for j in range(1, y-1):
            region = img[i-1:i+2, j-1:j+2]
            new = (region * kernel).sum()
            result[i,j] = new
    return result

```

```python
x, y = img.shape

c = OPS(x,y, kernel, result, img)

plt.figure(figsize=(12,10))

plt.subplot(1,2,1)
plt.imshow(img, cmap='grey')
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(c, cmap='grey')
plt.title('After Gauss')
plt.axis('off')
```

## Applying Sobel filter on the Image

```python
kernel = torch.tensor([ [-1,0,1],
                        [-2,0,2],
                        [-1,0,1]], dtype=torch.float32)

x, y = img.shape
result = torch.zeros_like(img) 
sobel1 = OPS(x, y, kernel, result, img)

plt.figure(figsize=(12,10))
plt.subplot(1,2,1)
plt.imshow(img, cmap='grey')
plt.title('Original')
plt.axis('off')


plt.subplot(1,2,2)
plt.imshow(sobel1, cmap='grey')
plt.title("sobel")
plt.axis('off')

```

## Same filter on horizontal direction 

```python
kernel = torch.tensor([ [-1,-2,-1],
                        [0,0,0],
                        [1,2,1]], dtype=torch.float32)

x, y = img.shape
result = torch.zeros_like(img) 
sobel2 = OPS(x, y, kernel, result, img)

plt.figure(figsize=(20,18))
plt.subplot(1,3,1)
plt.imshow(img, cmap='grey')
plt.title('Original')
plt.axis('off')


plt.subplot(1,3,2)
plt.imshow(sobel1, cmap='grey')
plt.title("sobel-vertical")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(sobel2, cmap='grey')
plt.title("sobel-horizontal")
plt.axis('off')
```

## Applying a lablasian Filter
- detects edges in all directions in one go 

```python
kernel = torch.tensor([
    [ 0,  1,  0],
    [ 1, -4,  1],
    [ 0,  1,  0]
], dtype=torch.float32)

x, y = img.shape
result = torch.zeros_like(img) 
sobel = OPS(x, y, kernel, result, img)

plt.figure(figsize=(12,10))
plt.subplot(1,2,1)
plt.imshow(img, cmap='grey')
plt.title('Original')
plt.axis('off')


plt.subplot(1,2,2)
plt.imshow(sobel, cmap='grey')
plt.title("Lablasian")
plt.axis('off')


```
